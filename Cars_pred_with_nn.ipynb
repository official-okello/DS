```python
import pandas as pd
import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import InputLayer, Dense, Normalization
from tensorflow.keras.losses import MeanAbsoluteError
from tensorflow.keras.metrics import RootMeanSquaredError
from tensorflow.keras.optimizers import Adam
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('train.csv')
```

**Data Preparation**

```python
tensor_data = tf.constant(data) # Converting the dataframe to a tensor for use during the model building
tensor_data.shape # Shape of the data does not change after conversion to tensor
```

```python
print(tensor_data)
```

```python
tensor_data = tf.cast(tensor_data, dtype=tf.float32) # Casting to float32 for compatibility and uniformity
tensor_data = tf.random.shuffle(tensor_data) # Shuffling the data for stochasticity
print(tensor_data)
```

```python
X = tensor_data[:, 3:-1] # From the 4th column to the second last columns
y = tensor_data[:, -1] # The last column

y = tf.expand_dims(y, axis=-1) # Expand the dims of the target so it becomes 1D. -1 for the last dimension.

print(X.shape)
print(y.shape)
```

```python
train_size = 0.8
val_size = 0.1
test_size = 0.1

X_train = X[:int(X.shape[0]*train_size)]
y_train = y[:int(y.shape[0]*train_size)]
train_dataset = tf.data.Dataset.from_tensor_slices((X_train, y_train))
train_dataset = train_dataset.shuffle(buffer_size=8, reshuffle_each_iteration=True).batch(32).prefetch(tf.data.AUTOTUNE)

X_val = X[int(X.shape[0]*train_size):int(X.shape[0]*(train_size + val_size))]
y_val = y[int(y.shape[0]*train_size):int(y.shape[0]*(train_size + val_size))]
val_dataset = tf.data.Dataset.from_tensor_slices((X_val, y_val))
val_dataset = val_dataset.shuffle(buffer_size=8, reshuffle_each_iteration=True).batch(32).prefetch(tf.data.AUTOTUNE)

X_test = X[int(X.shape[0]*(train_size + val_size)):]
y_test = y[int(y.shape[0]*(train_size + val_size)):]
test_dataset = tf.data.Dataset.from_tensor_slices((X_test, y_test))
test_dataset = test_dataset.shuffle(buffer_size=8, reshuffle_each_iteration=True).batch(32).prefetch(tf.data.AUTOTUNE)

print(f'X_train.shape:', X_train.shape)
print(f'y_train.shape:', y_train.shape)
print(f'X_val.shape:', X_val.shape)
print(f'y_val.shape:', y_val.shape)
print(f'X_test.shape:', X_test.shape)
print(f'y_test.shape:', y_test.shape)
```

```python
normalizer = Normalization() # Initializing the normalizer
normalizer.adapt(X_train) # Adapting the normalizer to the input tensor
normalizer(X) # Applying the normalizer to the input tensor
```

**Building the Neural Net**

```python
model = Sequential([
    InputLayer(input_shape=(8, )), # Input layer with 8 columns
    normalizer, # The normalization layer
    Dense(128, activation='relu'), # Hidden layer with 128 neurons and relu activation
    Dense(128, activation='relu'), # Hidden layer with 128 neurons and relu activation
    Dense(128, activation='relu'), # Hidden layer with 128 neurons and relu activation
    Dense(1) # Densen output layer with a single entity
])

model.summary() # Prints the summary of the model that includes layers, outputs and params
```

```python
tf.keras.utils.plot_model(model, to_file='model.png', show_shapes=True) # Plotting the model
```

```python
model.compile(
    loss=MeanAbsoluteError(),
    optimizer=Adam(learning_rate=0.1),
    metrics=[RootMeanSquaredError]) # Compiling with loss and optimizer Adam of lr 1.0(faster learning)
```

```python
history = model.fit(train_dataset, validation_data=val_dataset, epochs=200, verbose=0) # Training the model
```

```python
# Plot the losses
plt.plot(history.history['loss']) # Plots the losses from history.history
plt.plot(history.history['val_loss']) # Plots the validation losses from history.history
plt.title('Model Loss')
plt.ylabel('Loss')
plt.xlabel('Epoch')
plt.legend(['Train', 'Val'], loc='upper right') # Place the legend on the top right corner of the graph
plt.show()
```

```python
# Plot the metrics
plt.plot(history.history['root_mean_squared_error']) # Plots the rmse from history.history
plt.plot(history.history['val_root_mean_squared_error']) # Plots the validation rmse from history.history
plt.title('Model Performance')
plt.ylabel('RMSE')
plt.xlabel('Epoch')
plt.legend(['Train', 'Val'], loc='upper right') # Place the legend on the top right corner of the graph
plt.show()
```

```python
model.evaluate(X_test, y_test) # Evaluating the model
```
